```python
from time import sleep
import pandas as pd
import numpy as np
```

```python
from hanlp_restful import HanLPClient
def tokenizer(curpos):
	HanLP = HanLPClient('https://www.hanlp.com/api', auth=None, language='zh')   # auth不填则匿名，zh中文，mul多语种

	return HanLP.tokenize(curpos)
```

```python
datas = pd.read_csv("../datas/text_classify/train.csv", header=None, sep="\t")
datas
```

```python
corpus = datas.iloc[:,0]
corpus
```

```python
corpus = list(corpus)
corpus[0:200]
```

```python
corpus_list = []
for i in range(len(corpus)//200 + 1):
    corpus_split = tokenizer(corpus[i*200:(i+1)*200])
    corpus_list.append(corpus_split)
    sleep(31)
```

```python
corpus_list
```

```python
corpus_list_df = []
for i in corpus_list:
    corpus_list_df.extend(i)
len(corpus_list_df)
```

```python
new_data = pd.DataFrame([[' '.join(i) for i in corpus_list_df],list(datas.iloc[:,1])])
```

```python
new_data = new_data.T
```

```python
new_data.to_csv('./datas/text_classify/train_tokens.csv', index=False,header=False,sep='\t')
```

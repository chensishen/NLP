```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("../datas/text_classify/train_tokens.csv", header=None, sep="\t")
data
```

```python
corpus = data.iloc[:,0]
```

```python
all_words_list = []
corpus_split = []
for i in corpus:
    temp = i.split(' ')
    all_words_list.extend(temp)
    corpus_split.append(temp)
```

```python
all_words_set = set(all_words_list)
len(all_words_set)
```

```python
all_words_list2 = sorted(all_words_set)
all_words_list2.insert(0,'[UNK]')
all_words_list2.insert(0,'[PAD]')
all_words_list2
```

```python
vocab = dict(zip(all_words_list2, range(len(all_words_list2))))
len(vocab)
```

```python
vocab_df = pd.DataFrame([vocab.keys(),vocab.values()])
vocab_df.T.to_csv('./datas/text_classify/vocab.csv',header=False,index=False,sep='\t')
```

```python
def token2id(text,vocab):
    ids = [vocab.get(token,1) for token in text]
    return ids
```

```python
corpus_id = [token2id(text,vocab) for text in corpus_split]
corpus_id
```

```python
print(max([len(i) for i in corpus_id]))
print(min([len(i) for i in corpus_id]))
```

```python
corpus_maxlen = 30
from copy import deepcopy
corpus_id_padded = deepcopy(corpus_id)
for i in corpus_id_padded:
    if len(i)<corpus_maxlen:
        i.extend([0] * (corpus_maxlen-len(i)))
    elif len(i)>corpus_maxlen:
        i = i[:corpus_maxlen]
```

```python
print(corpus_id_padded[:10])
print(max([len(i) for i in corpus_id_padded]))
print(min([len(i) for i in corpus_id_padded]))
```

```python
corpus_id_df = pd.DataFrame(corpus_id_padded)
corpus_id_df.to_csv('./datas/text_classify/corpus_id.csv',header=False,index=False,sep='\t')
```
